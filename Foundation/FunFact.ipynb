```python
from openai import OpenAI
from IPython.display import Markdown, display

openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')

tellers = ["Seema", "Ruheen", "Mani", "Niya"]
facts = []
responses = []

def record (teller, fact):
    facts.append(fact)
    responses.append(f"Teller: {teller}\n\n{fact}")
    display(Markdown(f"Teller: {teller}\n\n{fact}"))

for teller in tellers:
    messages = [{"role": "user", "content": f"Your name is {teller}, tell an interesting fact on AI"}]
    response = openai.chat.completions.create(model="llama3.2", messages=messages)
    fact = response.choices[0].message.content
    record(teller, fact)

judge = f"""You are judging a competition between {len(tellers)} fact tellers. Each teller has been asked to tell an interesting fact on AI.

Your job is to evaluate thoroughly each response for interest, fun and factfullness of argument, and rank them in order of best to worst. 
You must also provide the reasoning behind you judgement

Respond with JSON, and only JSON, with the following format:
{{"results": ["best teller name":"reason", "second best teller name":"reason", "third best teller name":"reason", ...and so on]}}

Here are the responses from each competitor:

{responses}

Now respond with the JSON with the ranked order of the teller names, and the reason behind your judgement but nothing else. Do not include code blocks."""

judge_messages = [{"role": "user", "content": judge}]

judge_response = openai.chat.completions.create(model="llama3.2", messages=judge_messages)
outcome = judge_response.choices[0].message.content

display(Markdown(outcome))


```
